# Method dogfooding: hypersonic vehicle design

This notebook applies the four method patterns to ten concrete instances. The narrative records rationale, alternatives, uncertainty, and open issues rather than restating the vocabulary. The code cells are live OML/Compose content to copy into a description.

## Why this sequence

We start with ownership, then expose exchanges, then connect design obligations to stakeholders, and finally attach evidence. This order avoids creating orphan ports or tests. A flat component catalog would be faster, but it would not support subsystem-level tradeoffs. The examples intentionally keep free-form transfer text and scalar values: units, interface taxonomies, provenance, and result confidence are open issues.

```oml
// 1. System decomposition: aerodynamic and propulsion ownership
instance Sys_Aero : method:AerodynamicsAndStructures [ method:contains Part_Wings, Part_Nacelle ]
instance Sys_Propulsion : method:PropulsionSystem [ method:contains Part_Engine ]
```

```oml
// 2. System decomposition: thermal ownership
instance Sys_Thermal : method:ThermalSystem [ method:contains Part_NoseStackup, Part_LeadingEdgeStackup ]
```

```oml
// 3. Interface: aerodynamic airflow into propulsion
relation instance Conn_AeroToPropulsion : method:Connection [ from Port_Aero_Out to Port_Engine_In method:transfers "airflow" ]
```

```oml
// 4. Interface: control command loop
relation instance Conn_ControlLoop : method:Connection [ from Port_Stability_Command to Port_Aero_In method:transfers "control command" ]
```

## Review questions

The editors should flag a connection with reversed directions, a requirement without a complete trace, or evidence without a subsystem target. Before this method is used for certification, add units and provenance, replace transfer strings with typed interfaces, and model acceptance thresholds and result confidence.
